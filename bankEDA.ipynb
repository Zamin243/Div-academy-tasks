```python
dfc=df.copy() #hər bir ehtimala DataFramei kopyalayıb başqa bir variable`a mənimsədirəm.
```

```python
target_counts = df['deposit'].value_counts() #deposit sütununda olan hər bir unikal dəyərin neçə dəfə təkrarlandığını göstərir.
```

```python
plt.figure()
target_counts.plot(kind='bar', color=['red', 'green'])
plt.title('Target dəyişəninin paylanması (deposit)')
plt.xlabel('Abunə')
plt.ylabel('Sayı')
plt.show()
```

```python
plt.figure()
df.boxplot(column='age', by='deposit')
plt.title('Yaşın deposit üzrə paylanması')
plt.suptitle('')
plt.xlabel('Deposit')
plt.ylabel('Age')
plt.show()
```

```python
plt.figure()
df.boxplot(column='balance', by='deposit')
plt.title('Balansın deposit üzrə paylanması')
plt.suptitle('')
plt.xlabel('Deposit')
plt.ylabel('Balance')
plt.show()
```

```python
import seaborn as sns
numeric_cols = ['age', 'balance', 'duration', 'campaign']
for col in numeric_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

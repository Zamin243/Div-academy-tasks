```python
type_counts = df['type'].value_counts()
plt.figure(figsize=(6,4))
type_counts.plot(kind='bar', color=['blue', 'green'])
plt.title('Number of Movies vs TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

```python
rating_counts = df['rating'].value_counts()
plt.figure(figsize=(6,6))
rating_counts.plot(kind='pie', autopct='%1.1f%%', startangle=140, colors=['skyblue', 'orange', 'green'])
plt.title('Distribution of Netflix Ratings')
plt.ylabel('')
plt.show()
```

```python
titles_per_year = df['release_year'].value_counts().sort_index()
plt.figure(figsize=(8,5))
titles_per_year.plot(kind='line', marker='o', color='purple')
plt.title('Number of Netflix Titles Released per Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(True)
plt.show()
```

```python
top_countries = df['country'].value_counts().head(10)
plt.figure(figsize=(8,6))
top_countries.sort_values().plot(kind='barh', color='skyblue')
plt.title('Top 10 Countries with Most Netflix Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['release_year'], bins=range(df['release_year'].min(), df['release_year'].max()+2),
         color='skyblue', edgecolor='black')
plt.title('Distribution of Netflix Titles by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.xticks(range(df['release_year'].min(), df['release_year'].max()+1))
plt.grid(axis='y', linestyle='--')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='rating', order=df['rating'].value_counts().index, palette='pastel')
plt.title('Counts of Netflix Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.show()
```
